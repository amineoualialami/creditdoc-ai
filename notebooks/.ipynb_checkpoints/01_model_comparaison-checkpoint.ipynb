```python
import os
import time
import json
from pathlib import Path
from dotenv import load_dotenv
import litellm
import pandas as pd
from rich import print as rprint
from rich.table import Table
from rich.console import Console

load_dotenv("../.env")

MODELS_TO_COMPARE = [
    {"id": "ollama/llama3:8b", "label": "Llama 3 8B"},
    {"id": "ollama/qwen2.5:7b", "label": "Qwen 2.5 7B"},
    {"id": "ollama/phi4:latest", "label": "Phi-4 14B"},
]

OLLAMA_BASE = os.getenv("OLLAMA_API_BASE", "http://localhost:11434")

print(f"Models under test: {[m['label'] for m in MODELS_TO_COMPARE]}")
print(f"Ollama endpoint: {OLLAMA_BASE}")
```

```python
TEST_QUESTIONS = [
    # Definitional (easy baseline)
    {"id": "q1", "category": "definition",
     "question": "Quelle est la différence entre un TAEG et un TEG ?"},
    {"id": "q2", "category": "definition",
     "question": "Qu'est-ce que le ratio d'endettement maximum légal en France pour un prêt immobilier ?"},
    
    # Calculation / reasoning
    {"id": "q3", "category": "calculation",
     "question": "Pour un prêt de 10 000 EUR sur 24 mois à 4.5% TAEG, calcule la mensualité constante."},
    {"id": "q4", "category": "calculation",
     "question": "Un emprunteur gagne 3500 EUR nets par mois. Quel est le montant maximum de mensualités de crédit qu'il peut assumer selon la règle du 35% ?"},
    
    # Regulatory
    {"id": "q5", "category": "regulatory",
     "question": "Quelles sont les obligations d'information précontractuelle d'un prêteur en France pour un crédit à la consommation ?"},
    {"id": "q6", "category": "regulatory",
     "question": "Quelle est la durée minimale du droit de rétractation pour un crédit à la consommation en France ?"},
    
    # Nuanced / ambiguous (tests hallucination tendency)
    {"id": "q7", "category": "nuanced",
     "question": "Un dossier de crédit refusé peut-il être représenté auprès d'une autre banque sans mention de refus ?"},
    {"id": "q8", "category": "nuanced",
     "question": "Un co-emprunteur a-t-il les mêmes droits et obligations que l'emprunteur principal ?"},
    
    # Out-of-scope (tests honesty)
    {"id": "q9", "category": "out_of_scope",
     "question": "Quelle sera l'évolution du taux directeur de la BCE en 2027 ?"},
    {"id": "q10", "category": "out_of_scope",
     "question": "Devrais-je accepter cette offre de crédit de 15 000 EUR à 6% TAEG ?"},
]

print(f"{len(TEST_QUESTIONS)} questions across {len(set(q['category'] for q in TEST_QUESTIONS))} categories")
```

```python
SYSTEM_PROMPT = (
    "You are an assistant specialized in French consumer and mortgage credit. "
    "Answer in French, concisely and factually. "
    "If you don't know the answer or if it depends on future events / personal advice, "
    "say so explicitly instead of guessing."
)

def ask_model(model_id: str, question: str) -> dict:
    """Call a model via LiteLLM and capture answer + metrics."""
    start = time.time()
    try:
        response = litellm.completion(
            model=model_id,
            api_base=OLLAMA_BASE,
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": question},
            ],
            temperature=0.2,
        )
        latency_s = time.time() - start
        answer = response.choices[0].message.content
        usage = response.usage
        return {
            "model": model_id,
            "answer": answer,
            "latency_s": round(latency_s, 2),
            "prompt_tokens": usage.prompt_tokens,
            "completion_tokens": usage.completion_tokens,
            "error": None,
        }
    except Exception as e:
        return {
            "model": model_id,
            "answer": None,
            "latency_s": round(time.time() - start, 2),
            "prompt_tokens": 0,
            "completion_tokens": 0,
            "error": str(e),
        }
```

```python
results = []
console = Console()

for q in TEST_QUESTIONS:
    console.rule(f"[bold cyan]{q['id']} ({q['category']})[/]")
    print(f"Q: {q['question']}\n")
    for m in MODELS_TO_COMPARE:
        print(f"-> Asking {m['label']}...")
        r = ask_model(m["id"], q["question"])
        r["question_id"] = q["id"]
        r["question"] = q["question"]
        r["category"] = q["category"]
        r["model_label"] = m["label"]
        results.append(r)
        print(f"   latency: {r['latency_s']}s | tokens out: {r['completion_tokens']}")
        if r["error"]:
            print(f"   ERROR: {r['error']}")

# Save raw results
Path("results").mkdir(exist_ok=True)
with open("results/model_comparison_raw.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"\nDone — {len(results)} responses collected and saved.")
```

```python
df = pd.DataFrame(results)

summary = df.groupby("model_label").agg(
    avg_latency_s=("latency_s", "mean"),
    median_latency_s=("latency_s", "median"),
    avg_output_tokens=("completion_tokens", "mean"),
    total_errors=("error", lambda s: s.notna().sum()),
).round(2)

summary
```

```python
# Score each answer 1-5 across 4 dimensions.
# Fill this manually after reading cell 6 — this IS the exercise.
# 1 = unacceptable, 5 = excellent

SCORES = {
    # Format: (question_id, model_label): {dim: score}
    # Dimensions: factuality, completeness, concision, honesty_on_unknowns
    
    # Example (replace with your actual judgments):
    # ("q1", "Llama 3.3 8B"): {"factuality": 4, "completeness": 4, "concision": 3, "honesty_on_unknowns": 5},

    ("q1", "Llama 3 8B"): {"factuality": 2, "completeness": 2, "concision": 2, "honesty_on_unknowns": 5},
    ("q1", "Qwen 2.5 7B"): {"factuality": 1, "completeness": 1, "concision": 1, "honesty_on_unknowns": 5},
    ("q1", "Phi-4 14B"): {"factuality": 3, "completeness": 2, "concision": 4, "honesty_on_unknowns": 5},
    
    ("q2", "Llama 3 8B"): {"factuality": 4, "completeness": 3, "concision": 3, "honesty_on_unknowns": 5},
    ("q2", "Qwen 2.5 7B"): {"factuality": 4, "completeness": 4, "concision": 3, "honesty_on_unknowns": 5},
    ("q2", "Phi-4 14B"): {"factuality": 4, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},

    ("q3", "Llama 3 8B"): {"factuality": 3, "completeness": 2, "concision": 2, "honesty_on_unknowns": 5},
    ("q3", "Qwen 2.5 7B"): {"factuality": 3, "completeness": 2, "concision": 2, "honesty_on_unknowns": 5},
    ("q3", "Phi-4 14B"): {"factuality": 4, "completeness": 5, "concision": 4, "honesty_on_unknowns": 5},

    ("q4", "Llama 3 8B"): {"factuality": 5, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},
    ("q4", "Qwen 2.5 7B"): {"factuality": 5, "completeness": 2, "concision": 3, "honesty_on_unknowns": 5},
    ("q4", "Phi-4 14B"): {"factuality": 5, "completeness": 3, "concision": 3, "honesty_on_unknowns": 5},

    ("q5", "Llama 3 8B"): {"factuality": 4, "completeness": 4, "concision": 5, "honesty_on_unknowns": 5},
    ("q5", "Qwen 2.5 7B"): {"factuality": 4, "completeness": 4, "concision": 5, "honesty_on_unknowns": 5},
    ("q5", "Phi-4 14B"): {"factuality": 3, "completeness": 5, "concision": 3, "honesty_on_unknowns": 5},

    ("q6", "Llama 3 8B"): {"factuality": 5, "completeness": 4, "concision": 5, "honesty_on_unknowns": 5},
    ("q6", "Qwen 2.5 7B"): {"factuality": 1, "completeness": 2, "concision": 2, "honesty_on_unknowns": 5},
    ("q6", "Phi-4 14B"): {"factuality": 5, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},

    ("q7", "Llama 3 8B"): {"factuality": 5, "completeness": 3, "concision": 5, "honesty_on_unknowns": 5},
    ("q7", "Qwen 2.5 7B"): {"factuality": 3, "completeness": 2, "concision": 3, "honesty_on_unknowns": 5},
    ("q7", "Phi-4 14B"): {"factuality": 2, "completeness": 3, "concision": 2, "honesty_on_unknowns": 5},

    ("q8", "Llama 3 8B"): {"factuality": 3, "completeness": 2, "concision": 2, "honesty_on_unknowns": 5},
    ("q8", "Qwen 2.5 7B"): {"factuality": 2, "completeness": 3, "concision": 3, "honesty_on_unknowns": 5},
    ("q8", "Phi-4 14B"): {"factuality": 4, "completeness": 4, "concision": 4, "honesty_on_unknowns": 5},

    ("q9", "Llama 3 8B"): {"factuality": 5, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},
    ("q9", "Qwen 2.5 7B"): {"factuality": 5, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},
    ("q9", "Phi-4 14B"): {"factuality": 5, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},

    ("q10", "Llama 3 8B"): {"factuality": 4, "completeness": 5, "concision": 5, "honesty_on_unknowns": 5},
    ("q10", "Qwen 2.5 7B"): {"factuality": 4, "completeness": 5, "concision": 4, "honesty_on_unknowns": 5},
    ("q10", "Phi-4 14B"): {"factuality": 4, "completeness": 2, "concision": 2, "honesty_on_unknowns": 3},

    
}

# Build the scoring DataFrame
scoring_rows = []
for (qid, model), dims in SCORES.items():
    row = {"question_id": qid, "model_label": model, **dims}
    scoring_rows.append(row)

if not scoring_rows:
    print("SCORES dict is empty — fill it above with your judgments.")
    scores_summary = None
    scores_by_category = None
else:
    scores_df = pd.DataFrame(scoring_rows)
    scores_df["total"] = scores_df[DIMENSIONS].sum(axis=1)

    # 1) Per-model averages across all questions
    scores_summary = scores_df.groupby("model_label")[DIMENSIONS + ["total"]].mean().round(2)

    # 2) Per-model x per-category average total (saved for cell 9)
    scores_df_with_cat = scores_df.merge(
        df[["question_id", "category"]].drop_duplicates(),
        on="question_id",
        how="left",
    )
    scores_by_category = (
        scores_df_with_cat
        .groupby(["model_label", "category"])["total"]
        .mean()
        .round(2)
        .unstack()
    )

    # Reorder categories in a logical sequence (easy to hard)
    category_order = ["definition", "calculation", "regulatory", "nuanced", "out_of_scope"]
    existing = [c for c in category_order if c in scores_by_category.columns]
    scores_by_category = scores_by_category[existing]

    print(f"Scored {len(scoring_rows)} / {len(TEST_QUESTIONS) * len(MODELS_TO_COMPARE)} possible pairs.\n")
    print("=== Per-model average scores (all questions) ===")
    display(scores_summary)
    print("\n=== Per-model x category — average TOTAL per question (max 20) ===")
    display(scores_by_category)
```
